```python
# Numerical outliers

outlier_summary = []

for col in numerical_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ).sum()

    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outliers": outliers,
        "Outlier %": (outliers / len(df)) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.sort_values(
    by="Outliers",
    ascending=False
)
```

```python
# Suspicious negative values

negative_summary = []

for col in numerical_columns:

    negative_count = (df[col] < 0).sum()

    negative_summary.append({
        "Column": col,
        "Negative Values": negative_count
    })

negative_summary = pd.DataFrame(negative_summary)

negative_summary[
    negative_summary["Negative Values"] > 0
]
```

```python
# Delivery performance by shipping mode

df["Shipping Mode"].value_counts()
```

```python
shipping_mode_performance = (
    df.groupby("Shipping Mode")
      .agg(
          Orders=("Shipping Mode", "size"),
          Avg_Actual_Days=(
              "Days for shipping (real)",
              "mean"
          ),
          Avg_Scheduled_Days=(
              "Days for shipment (scheduled)",
              "mean"
          ),
          Late_Risk_Rate=(
              "Late_delivery_risk",
              "mean"
          )
      )
      .reset_index()
)

shipping_mode_performance
```

```python
shipping_mode_performance["Late_Risk_Rate"] *= 100

shipping_mode_performance
```

```python
# Average actual shipping time by shipping mode

plt.figure(figsize=(10, 5))

sns.barplot(
    data=shipping_mode_performance,
    x="Shipping Mode",
    y="Avg_Actual_Days"
)

plt.title("Average Actual Shipping Time by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("Average Shipping Time (Days)")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

```python
# Late deliivery risk by shipping mode

plt.figure(figsize=(10, 5))

sns.barplot(
    data=shipping_mode_performance,
    x="Shipping Mode",
    y="Late_Risk_Rate"
)

plt.title("Late Delivery Risk by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("Late Delivery Risk (%)")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```

```python
# Delivery performance by market

market_performance = (
    df.groupby("Market")
      .agg(
          Orders=("Market", "size"),
          Avg_Actual_Days=(
              "Days for shipping (real)",
              "mean"
          ),
          Avg_Scheduled_Days=(
              "Days for shipment (scheduled)",
              "mean"
          ),
          Late_Risk_Rate=(
              "Late_delivery_risk",
              "mean"
          )
      )
      .reset_index()
)

market_performance["Late_Risk_Rate"] *= 100

market_performance
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=market_performance,
    x="Market",
    y="Late_Risk_Rate"
)

plt.title("Late Delivery Risk by Market")
plt.xlabel("Market")
plt.ylabel("Late Delivery Risk (%)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Delivery performnce by region

region_performance = (
    df.groupby("Order Region")
      .agg(
          Orders=("Order Region", "size"),
          Avg_Actual_Days=(
              "Days for shipping (real)",
              "mean"
          ),
          Late_Risk_Rate=(
              "Late_delivery_risk",
              "mean"
          )
      )
      .reset_index()
)

region_performance["Late_Risk_Rate"] *= 100

region_performance.sort_values(
    "Late_Risk_Rate",
    ascending=False
).head(10)
```

# **KPI TABLE**

```python
delivery_kpis = {
    "Total Orders": float(len(df)),
    "Late Delivery Risk Rate (%)": float(df["Late_delivery_risk"].mean() * 100),
    "Average Actual Shipping Time (Days)": float(df["Days for shipping (real)"].mean()),
    "Average Scheduled Shipping Time (Days)": float(df["Days for shipment (scheduled)"].mean()),
    "Schedule Breach Rate (%)": (
        (
            df["Days for shipping (real)"]
            >
            df["Days for shipment (scheduled)"]
        ).mean() * 100
    )
}

delivery_kpis
```
